```python
import numpy as np

from exercises.src.e01_maxent_discrete import maxent_from_moments
```

```python
# Example 1: 约束‘平均点数’（骰子）
x = np.arange(1, 7)
F = x.reshape(-1, 1)

# 改变约束值 μ=E[x]，观察 λ 如何变化
for mu in [2.0, 3.0, 3.5, 4.0, 5.0]:
    res = maxent_from_moments(F, np.array([mu]))
    lam = float(res.lambdas[0])
    p = res.p
    print(f"μ={mu:>4.1f}  λ={lam:>8.4f}  E_p[x]={p @ x:>6.3f}  p={np.round(p, 3)}")
```
